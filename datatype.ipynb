SETS
a set is an unordered collection of items . every set element is 
unqiue and must be immutable .However , a set itself is mutable . We can add or remove items from it.


```python
s1={}
s=set()
print(type(s1))
print('empty set ',type(s))
```

```python
# set operation
# Union 
s1={1,2,4,5,6,7}
s2={2,4,1,3,52,4,56,2,3}
s1 | s2
# Intersection
s1&s2
# difference
s1-s2
# symmetric difference(^)
s1^s2
# membership test()  in and not 
1 in s1
# iteration
for i in s1:
    print(i)
```

```python
#set function len/min/max/sum/sorted 
s={1,25,6,4,62,57,85,8,53,2,42,34,}
len(s)
min(s)
max(s)

print(sum(s))
sorted(s)
```

```python
# union /update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.union(s2)
s1.update(s2) 
# in update s2 element are copy in s1
print(s1,s2)
```

```python
# intersection/intersection_update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.intersection(s2)
s1.intersection_update(s2) # copy intersection past it into s1

print(s2)
```
